```python
# importar bibliotecas
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import accuracy_score
```

Carregamento e análise dos dados

```python
# carregar dataset
df = pd.read_csv('/datasets/users_behavior.csv')

# visualizar primeiras linhas
print(df.head())

# informações gerais
print(df.info())

# verificar valores ausentes
print(df.isna().sum())

# verificar duplicados
print(df.duplicated().sum())
```

<div class="alert alert-block alert-success">
<b> Comentário do revisor: </b> <a class="tocSkip"></a>

- O conjunto de dados foi carregado corretamente
- As bibliotecas necessárias foram importadas corretamente.
</div>

Conclusão da análise inicial

- O conjunto de dados já está pré-processado.
- Não há valores ausentes.
- Não existem duplicatas.
- A coluna alvo é is_ultra.
- As demais colunas serão usadas como características

Separação dos dados
Vamos dividir os dados em:

- 60% treinamento
- 20% validação
- 20% teste

```python
# separar características e alvo
features = df.drop(['is_ultra'], axis=1)
target = df['is_ultra']

# separar treino (60%) e restante (40%)
features_train, features_temp, target_train, target_temp = train_test_split(
    features,
    target,
    test_size=0.4,
    random_state=12345
)

# dividir restante em validação (20%) e teste (20%)
features_valid, features_test, target_valid, target_test = train_test_split(
    features_temp,
    target_temp,
    test_size=0.5,
    random_state=12345
)

print(features_train.shape)
print(features_valid.shape)
print(features_test.shape)
```

Justificativa da divisão
- O conjunto de treinamento precisa ser maior para o modelo aprender melhor.
- Os conjuntos de validação e teste possuem tamanhos iguais para avaliação justa.
- O conjunto de validação será usado para escolher hiperparâmetros.
- O conjunto de teste será usado apenas na avaliação final.

<div class="alert alert-block alert-success">
<b> Comentário do revisor: </b> <a class="tocSkip"></a>

- Os conjuntos de dados foram divididos em `treino`, `validação` e `teste`.
- Os conjuntos de dados foram dividos em tamanhos adequados.
</div>

Modelo Árvore de Decisão

```python
best_model_tree = None
best_result_tree = 0
best_depth = 0

for depth in range(1, 21):

    model = DecisionTreeClassifier(
        random_state=12345,
        max_depth=depth
    )

    model.fit(features_train, target_train)

    predictions_valid = model.predict(features_valid)

    accuracy = accuracy_score(target_valid, predictions_valid)

    if accuracy > best_result_tree:
        best_result_tree = accuracy
        best_depth = depth
        best_model_tree = model

print("Melhor acurácia:", best_result_tree)
print("Melhor profundidade:", best_depth)
```

A árvore de decisão apresentou bom desempenho, porém tende a sofrer overfitting em profundidades maiores.

Modelo Random Forest

```python
best_model_forest = None
best_result_forest = 0
best_est = 0
best_depth = 0

for est in range(10, 101, 10):

    for depth in range(1, 16):

        model = RandomForestClassifier(
            random_state=12345,
            n_estimators=est,
            max_depth=depth
        )

        model.fit(features_train, target_train)

        predictions_valid = model.predict(features_valid)

        accuracy = accuracy_score(target_valid, predictions_valid)

        if accuracy > best_result_forest:
            best_model_forest = model
            best_result_forest = accuracy
            best_est = est
            best_depth = depth

print("Melhor acurácia:", best_result_forest)
print("Melhor n_estimators:", best_est)
print("Melhor max_depth:", best_depth)
```

O modelo Random Forest apresentou a maior acurácia entre os modelos testados, mantendo melhor equilíbrio entre viés e variância.

Modelo Regressão Logística

```python
model_lr = LogisticRegression(random_state=12345, max_iter=1000)

model_lr.fit(features_train, target_train)

predictions_valid = model_lr.predict(features_valid)

accuracy_lr = accuracy_score(target_valid, predictions_valid)

print("Acurácia da Regressão Logística:", accuracy_lr)
```

A regressão logística apresentou desempenho inferior aos modelos baseados em árvores

Comparação dos modelos

| Modelo              | Resultado |
| ------------------- | --------- |
| Decision Tree       | ~0.78     |
| Random Forest       | ~0.81     |
| Logistic Regression | ~0.74     |

O modelo Random Forest apresentou a melhor acurácia no conjunto de validação.

Teste final do modelo

```python
# usar melhor modelo
final_model = RandomForestClassifier(
    random_state=12345,
    n_estimators=best_est,
    max_depth=best_depth
)

# treinar novamente
final_model.fit(features_train, target_train)

# previsões no teste
predictions_test = final_model.predict(features_test)

# calcular acurácia
final_accuracy = accuracy_score(target_test, predictions_test)

print("Acurácia no conjunto de teste:", final_accuracy)
```

Prova real do modelo

```python
# previsões usando dados de treino
train_predictions = final_model.predict(features_train)

# acurácia treino
train_accuracy = accuracy_score(target_train, train_predictions)

print("Acurácia treino:", train_accuracy)
print("Acurácia teste:", final_accuracy)
```

Conclusão da prova real
- A acurácia de treino e teste ficaram relativamente próximas.
- Isso indica que o modelo generaliza bem.
- Não foram observados sinais fortes de overfitting.
